### Make your Jupyter Notebook capable of running asynchronous functions.

🧠 import nest_asyncio and nest_asyncio.apply() — What does it mean?\
It’s used to fix issues with Python's asyncio event loop, especially in environments like Jupyter notebooks or when you’re running async code inside code that’s already running async.

💡 Breakdown:\
🔹 asyncio is Python's library for asynchronous programming (e.g., running multiple tasks at once).
But Python allows only one event loop per thread, and sometimes you get this kind of error:

### RuntimeError: This event loop is already running
This happens in notebooks, or when an AI agent calls async functions inside already running async code.

✅ What nest_asyncio does:\
It patches the running event loop to allow it to be nested.

So now you can run async code inside async code without error.

Especially useful when you're using:

asyncio.run() in a Jupyter notebook

LangChain agents with tools

FastAPI inside another async context

🔧 How to use it:

import nest_asyncio\
nest_asyncio.apply()

That’s it! After this, you can safely run await and async functions without that annoying RuntimeError.


🚀 Real-world uses in Agentic AI:\
LangChain agents calling tools functions

Agents that run event loops inside event loops


```python
import nest_asyncio
nest_asyncio.apply()
```

## How to configure LLM Providers at different levels (Global, Run and Agent)?

Agents SDK is setup to use OpenAI as default providers. When using other providers you can setup at different levels:
1. Agent Level
2. RUN LEVEL
3. Global Level

We will always your Agent Level Configuration so each agent can use the LLM best fit for it.

### 1. AGENT LEVEL

```python
import asyncio
from openai import AsyncOpenAI
from agents import Agent, OpenAIChatCompletionsModel, Runner


#Reference: https://ai.google.dev/gemini-api/docs/openai
client = AsyncOpenAI(
    api_key=gemini_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)

async def main():
    # This agent will use the custom LLM provider
    agent = Agent(
        name="Assistant",
        instructions="You only respond in urdu.",
        model=OpenAIChatCompletionsModel(model="gemini-2.5-flash", openai_client=client),
    )

    result = await Runner.run(
        agent,
        "I am learning Agentic AI with Panaversity Community",
    )
    print(result.final_output)


if __name__ == "__main__":
    asyncio.run(main())
```

### 2. RUN LEVEL

```python
from agents import Agent, Runner, AsyncOpenAI, OpenAIChatCompletionsModel
from agents.run import RunConfig

#Reference: https://ai.google.dev/gemini-api/docs/openai
external_client = AsyncOpenAI(
    api_key=gemini_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)

model = OpenAIChatCompletionsModel(
    model="gemini-2.0-flash",
    openai_client=external_client
)

config = RunConfig(
    model=model,
    model_provider=external_client,
    tracing_disabled=True
)

agent: Agent = Agent(name="Assistant", instructions="You are a helpful assistant")

result = Runner.run_sync(agent, "Hello, how are you.", run_config=config)

print(result.final_output)
```

### GLOBAL
