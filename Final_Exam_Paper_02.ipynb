Q#3- Create three functions: addition, subtraction, and division. Register these
functions as tools that an agent can use.
Pass the input “What is 2 + 2?” to your agent. The agent should detect that
addition is needed and use the addition tool to return the correct result.

Step 1: Define the functions

I have created three simple functions: addition, subtraction, and division.

Step 2: Register the functions as tools

I represent the tools in a dictionary so the agent can access them by name.

Step 3: Build the agent

The agent looks at the user's input, decides which operation is needed, and calls the correct tool.

```python
#Import Required Libraries
import os
from dotenv import load_dotenv

from agents import (
    Agent, Runner,
    AsyncOpenAI, OpenAIChatCompletionsModel,
    function_tool, set_tracing_disabled,
)

# Load environment variables
# load_dotenv() # Not needed when using Colab userdata

from google.colab import userdata
# userdata.get('GEMINI_API_KEY') # We will use OPENAI_API_KEY for clarity

# Disable tracing (optional, for clean output)
set_tracing_disabled(disabled=True)

# 1) Environment & Client Setup
# Use the Colab secrets manager for API keys
OPENAI_API_KEY = userdata.get('GEMINI_API_KEY') # Get the key from Colab secrets and name it as expected

BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

external_client: AsyncOpenAI = AsyncOpenAI(
    api_key=OPENAI_API_KEY, # Pass the key directly
    base_url=BASE_URL,
)

#  2) Model Initialization
model: OpenAIChatCompletionsModel = OpenAIChatCompletionsModel(
    model="gemini-2.5-flash",
    openai_client=external_client
)

#  3) Define tools (functions)
@function_tool
def addition(a: int, b: int) -> int:
    """ Add two numbers."""
    return a + b

@function_tool
def subtraction(a: int, b: int) -> int:
    """ Subtract b from a."""
    return a - b

@function_tool
def division(a: int, b: int) -> float:
    """ Divide a by b (with zero-check)."""
    if b == 0:
        return "Error: Division by zero"
    return a / b

# 4) Create agent and register tools
agent: Agent = Agent(
    name="MathAssistant",
    instructions="You are a helpful math agent. Always use tools for calculations.",
    model=model,
    tools=[addition, subtraction, division],
)

# Instead of run_sync, use run (async)
result = await Runner.run(agent, prompt)
print(result.final_output)

#  Print final result
print("\n CALLING AGENT\n")
print(result.final_output)
```
