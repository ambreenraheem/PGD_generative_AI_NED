app.py

```python
import gradio as gr
import google.generativeai as genai
import os

# Load Gemini API
genai.configure(api_key=os.getenv("GEMINI_API_KEY"))

SYSTEM_PROMPT = """
You are an AI Email Composer Agent.
Compose clear, professional emails with:
- Subject
- Greeting
- Body
- Closing
- Signature
"""

def compose_email(user_input):
    response = genai.GenerativeModel("gemini-2.5-flash").generate_content(
        [SYSTEM_PROMPT, user_input]
    )
    return response.text

iface = gr.Interface(
    fn=compose_email,
    inputs=gr.Textbox(lines=3, placeholder="Write: Email HR for leave on Friday"),
    outputs="text",
    title="AI Email Assistant",
    description="Generates professional email drafts"
)

iface.launch()
```
