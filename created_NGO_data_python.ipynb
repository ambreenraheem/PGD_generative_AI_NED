```python
import pandas as pd
import random
from faker import Faker

# Initialize Faker
fake = Faker()

# Records
n_records = 6000

# Date range 2021–2025
date_range = pd.date_range("2021-01-01", "2025-12-31")

# Fraud types and indicators
fraud_types = {
    "Grant Application Fraud": [
        "Inflated budget, fake NGO registration",
        "Misrepresented credentials",
        "Fake applicants with forged documents"
    ],
    "Expense/Reimbursement Fraud": [
        "Fake invoices submitted",
        "Duplicate travel bills",
        "Over-reporting operational costs"
    ],
    "Procurement & Vendor Fraud": [
        "Kickbacks from vendor",
        "Inflated vendor rates",
        "Related-party transaction with NGO collusion"
    ],
    "Payroll & HR Fraud": [
        "Ghost employees on payroll",
        "Inflated salaries for relatives",
        "Same employee on multiple grants"
    ],
    "Misuse of Funds": [
        "Funds diverted to personal account",
        "Project not executed as reported",
        "Grant used for unrelated purposes"
    ],
    "Collusion / Conflict of Interest": [
        "Awarded contract to family-owned company",
        "Employee favored friend's company",
        "Beneficiary selection biased towards relatives"
    ],
    "None": ["Genuine"]
}

fraud_type_list = list(fraud_types.keys())

# Cities
cities = [
    "Karachi", "Lahore", "Islamabad", "Rawalpindi", "Peshawar", "Quetta",
    "Faisalabad", "Multan", "Hyderabad", "Sialkot", "Sukkur", "Gujranwala",
    "Bahawalpur", "Mardan", "Abbottabad"
]

# Generate 120 unique NGO names
ngo_names = [fake.company() + " Foundation" for _ in range(120)]

# Generate 120 unique Project Titles
project_titles = [
    "Solar Panels for Schools", "Maternal Health Clinics", "Food Distribution",
    "Women Skill Development", "Clean Water Access", "Shelter for Flood Victims",
    "Child Nutrition Program", "Vocational Training for Women", "COVID Relief Packages",
    "Literacy Project", "Microfinance for Farmers", "Orphan Support Project",
    "Free Medical Camps", "Youth IT Training", "Disaster Relief Operations",
    "Green Energy for Villages", "Road Safety Awareness", "Waste Recycling Project",
    "Scholarship Program", "Rural Electrification", "Digital Literacy for Girls",
    "Small Business Grants", "Tree Plantation Drive", "Mobile Health Units",
    "Housing for Homeless", "Hygiene Awareness", "School Renovation Project",
    "Community Development Program", "Drug Rehabilitation Center", "Street Children Shelter"
]

while len(project_titles) < 120:
    project_titles.append(fake.catch_phrase())

# Generate records
records = []
for i in range(1, n_records + 1):
    record_id = f"PKF{i:05d}"
    date_val = random.choice(date_range)
    applicant_name = fake.name()
    ngo_name = random.choice(ngo_names)
    project_title = random.choice(project_titles)
    city = random.choice(cities)
    requested_amount = random.randint(500000, 30000000)
    legit_amount = requested_amount - random.randint(0, int(requested_amount * 0.6))

    fraud_type = random.choice(fraud_type_list)
    fraud_flag = 0 if fraud_type == "None" else 1
    fraud_indicator = random.choice(fraud_types[fraud_type])

    records.append([
        date_val, record_id, applicant_name, ngo_name, city, project_title,
        requested_amount, legit_amount, fraud_flag, fraud_type, fraud_indicator
    ])

# Create DataFrame
df = pd.DataFrame(records, columns=[
    "Date", "Record_ID", "Applicant_Name", "NGO_Name", "City", "Project_Title",
    "Requested_Amount (PKR)", "Legitimate_Estimate (PKR)",
    "Fraud_Flag", "Fraud_Type", "Fraud_Indicator"
])

# Save CSV
df.to_csv("pakistan_ngo_fraud_dataset_6000_full.csv", index=False)
print("✅ Dataset generated: pakistan_ngo_fraud_dataset_6000_full.csv")
```
