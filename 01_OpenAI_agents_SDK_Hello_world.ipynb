# 📦 Imports

🧠 import nest_asyncio and nest_asyncio.apply() — What does it mean?
It’s used to fix issues with Python's asyncio event loop, especially in environments like Jupyter notebooks or when you’re running async code inside code that’s already running async.

💡 Breakdown:\
🔹 asyncio is Python's library for asynchronous programming (e.g., running multiple tasks at once). But Python allows only one event loop per thread, and sometimes you get this kind of error:

### RuntimeError: This event loop is already running
This happens in notebooks, or when an AI agent calls async functions inside already running async code.

✅ What nest_asyncio does:\
It patches the running event loop to allow it to be nested.

So now you can run async code inside async code without error.

Especially useful when you're using:

asyncio.run() in a Jupyter notebook

LangChain agents with tools

FastAPI inside another async context

🔧 How to use it:

import nest_asyncio\
nest_asyncio.apply()

That’s it! After this, you can safely run await and async functions without that annoying RuntimeError.

🚀 Real-world uses in Agentic AI:\
LangChain agents calling tools functions

Agents that run event loops inside event loops

# 💬 Step 3: Running Agent Asynchronously

### You're now using asynchronous execution (async/await) to run your agent in a non-blocking way. This is important when you're dealing with LLMs, tools, or web apps that require efficient task management.

🔹 import asyncio
- You’re importing Python’s built-in asyncio library.
- This lets you run asynchronous code (non-blocking, fast, efficient).
- It’s often used in Agentic AI, APIs, or tools like FastAPI, LangChain, etc.

🔹 async def main():
- Defines an asynchronous function named main.
- Inside this, you can use await to pause execution until a task completes — like calling an LLM or fetching a web API.

🔹 await Runner.run(...)
- result: Runner = await Runner.run(
    math_agent, "Tell me about recursion in programming.")

- You’re calling the agent, but using await means:

- Let the program continue doing other things while it waits for the response.

- This is helpful when calling LLMs (like Gemini or GPT-4), which might take 1–5 seconds to respond.

🔹 asyncio.run(main())
- This line starts the async event loop and runs your main() function.
- Without this, your async function won’t execute.
- Only one asyncio.run(...) can be active at a time — which is why we sometimes use nest_asyncio in notebooks.

```python
import asyncio

async def main():

    result: Runner = await Runner.run(math_agent, "Tell me about recursion in programming.")

    print(result.final_output)

async def get_recursion_explanation():
    result: Runner = await Runner.run(math_agent, "Tell me about recursion in programming.")
    return result.final_output


asyncio.run(main())
```

## Example 1:
### 👨‍🍳🍽️ Recipe Bot

```python
set_tracing_disabled(disabled=True)

# Client Setup for Connecting to Gemini
external_client:AsyncOpenAI = AsyncOpenAI(
    api_key=gemini_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)

#Initialize model
model:OpenAIChatCompletionsModel = OpenAIChatCompletionsModel(
    model="gemini-2.5-flash",
    openai_client=external_client
)

def main():
  # Create the Recipe Agent
  agent = Agent(
      name="RecipeBot",
      instructions=(
          """You are a helpful recipe assistant. A user will give you a few ingredients
          they have at home, and you will suggest one simple and quick recipe using only those items.
          Keep it short, step-by-step, and easy for beginners to cook."""
      ),
      model=model
  )

  print("\n🍳 What can I cook today?\n")
  ingredients = "eggs, tomatoes, onions, and bread"
  result:Runner = Runner.run_sync(agent, f"I have these at home: {ingredients}. What can I cook?")

  print(result.final_output)
  return result
```

```python
if __name__ == "__main__":
    main()
```
