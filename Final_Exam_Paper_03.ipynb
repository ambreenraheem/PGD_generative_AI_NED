```python
from mcp.server.fastmcp import FastMCP, tool
import uvicorn
import nest_asyncio

# Fix event loop issues in notebooks
nest_asyncio.apply()

#  1) Define MCP server (stateless)
mcp = FastMCP("hello-server")

# 2) Define a tool
@tool
def say_hello() -> str:
    """Returns a Hello World message."""
    return "Hello, World!"

# 3) Create HTTP app (use asgi_app instead of streamable_http_app)
app = mcp.asgi_app()

#4) Run the server using uvicorn
uvicorn.run(app, host="0.0.0.0", port=8000)
```
