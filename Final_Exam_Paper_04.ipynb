```python
import os
from openai import OpenAI

from google.colab import userdata

# 🔑 Gemini API setup
client = OpenAI(
    api_key=userdata.get('GEMINI_API_KEY'),   # Pass the API key directly
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)


# 🧑‍💼 Manager Agent
def manager_agent(task: str) -> str:
    response = client.chat.completions.create(
        model="gemini-2.5-flash",  # ✅ using Gemini Flash
        messages=[
            {"role": "system", "content": "You are the Manager. Refine and clarify user tasks before assigning them."},
            {"role": "user", "content": task}
        ]
    )
    return response.choices[0].message.content

# 💻 Web Developer Agent
def web_dev_agent(refined_task: str) -> str:
    response = client.chat.completions.create(
        model="gemini-2.5-flash",
        messages=[
            {"role": "system", "content": "You are a Web Developer. Convert the Manager's refined task into technical website requirements."},
            {"role": "user", "content": refined_task}
        ]
    )
    return response.choices[0].message.content

# 📱 Mobile App Developer Agent
def mobile_dev_agent(web_task: str) -> str:
    response = client.chat.completions.create(
        model="gemini-2.5-flash",
        messages=[
            {"role": "system", "content": "You are a Mobile App Developer. Extend the Web Developer's work into a mobile app and provide the final output."},
            {"role": "user", "content": web_task}
        ]
    )
    return response.choices[0].message.content

# 🚀 Workflow Orchestration
def run_workflow(user_input: str):
    print("📥 User Input:", user_input)

    refined = manager_agent(user_input)
    print("\n🧑‍💼 Manager Refined Task:\n", refined)

    web_result = web_dev_agent(refined)
    print("\n💻 Web Developer Output:\n", web_result)

    final_output = mobile_dev_agent(web_result)
    print("\n📱 Mobile App Developer Final Output:\n", final_output)
    return final_output # Return the final output

# ▶️ Example Run
if __name__ == "__main__":
    run_workflow("Create a student management website for me.")
```

```python
# For presentable result in Markdown
from IPython.display import Markdown

final_output = run_workflow("Create a student management website for me.")
Markdown(final_output)
```
