Q#1- How many primitives are defined in the OpenAI Agents SDK, and what does each one mean? Please explain in detail with real-world examples.

Ans # 1:

Understanding the OpenAI Agent SDK
OpenAI Agent SDK is a framework that helps you build agentic AI apps with an easy-to-use and lightweight package. The Agents SDK has the following primitives:

01- Agents: LLMs equipped with predefined instructions and integrated tools.

Example: Imagine I am building a virtual homework assistant app. I might create:\
- A Math Tutor Agent that solves math problems,
- A History Tutor Agent that explains historical events,
- A Triage Agent that figures out whether the user's query should go to the math tutor or history tutor.
- The Triage Agent inspects the user's question ("Who was the first president of the United States?" or "Solve 2x + 3 = 7"). It then delegates to the correct expert agent.

02- Handoffs: Allow agents to communicate and delegate tasks to one another.\
Example: Continuing with our homework assistant, say the user asks a mixed question: "Explain the causes of World War II and help me solve a geometry problem." The Triage Agent can determine that it's best to hand off to two separate agents: the History Agent and the Math Agent.
This allows each specialized agent to focus on its own domain.

03- Guardrails: Ensure that inputs to agents are properly validated before processing.\
Example: Suppose I ask the Math Agent to solve: "Who is the Founder of Pakistan?" That's a questionable scenario. A guardrail might check for inappropriate or nonsensical requests and stop the agent from proceeding—or raise an exception to handle it more gracefully.

04- Sessions: Sessions automatically manage conversation history across runs of agents and same "Function:" They act like memory—agents don't start from scratch each time; they remember what happened before.
Example: If I ask first to the tutor agents: "What's Decimal number?" and they explain it.after my "Okay, and what about Whole Number?" The session maintains the context so agents understand that you're still talking about derivatives.

01- Trigger No0des:Start workflows based on events or schedules.\

02- Action-Integration Nodes: Perform external or internal operations (via APIs, DBs, emails)\
Use of Action: Fundamental task executors—send, fetch, update, etc.\
03- Core Nodes: Generic logic, data transformation, branching.\
Use of Core Nodes: Manipulate, split, or consolidate data as needed.\
04- Output Nodes: Deliver results emails, webhook responses, notifications.\
Use of Output Nodes: Final outputs—notify users or confirm actions,.\
05- Expressions & Variables: Dynamic data insertion and configuration.
Use of Expression and Variables: Personalize and parameterize workflow behavior.\
06- Credentials: Secure storage of sensitive access details.\
Use of Credentials: Secure, reusable access for API-enabled nodes.\
07- Modularity (Sub-Workflows): Reuse workflow segments by calling them from other workflows.\
Use of Modularity:Break down complex automations into manageable parts.\
08- Cluster Nodes: Composite nodes with root + sub-node structures, often for advanced use-cases (e.g., AI)\
